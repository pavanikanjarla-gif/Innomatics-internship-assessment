# Problem 1: Social Media – Post Engagement Analyzer

Problem Statement

Create a function to analyze engagement on a social media post.


```python
# Function to analyze post engagement
def analyze_post_engagement(likes_list):
    
    total_likes = 0
    
    # Step 1: Calculate total likes using loop
    for like in likes_list:
        total_likes += like
    
    print("Total Likes:", total_likes)
    
    # Step 2: Check post status
    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")


# Example data (likes from different platforms or time slots)
likes = [250, 300, 150, 200, 150]

# Call function
analyze_post_engagement(likes)
```

# Problem 3: Agriculture – Rainfall Adequacy Checker


Problem Statement

Create a function that analyzes seasonal rainfall data.


```python
# Function to check rainfall adequacy
def check_rainfall(rainfall_data, required_level):
    
    total_rainfall = 0
    
    # Step 1: Calculate total rainfall using loop
    for value in rainfall_data:
        total_rainfall += value
    
    # Step 2: Calculate average rainfall
    average_rainfall = total_rainfall / len(rainfall_data)
    
    print("Average Rainfall:", int(average_rainfall))
    
    # Step 3: Check adequacy
    if average_rainfall >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")


# Example seasonal rainfall data
rainfall = [70, 75, 80, 65, 70]   # average = 72
required = 70

# Call function
check_rainfall(rainfall, required)
```

# Problem 4: Social Media – Duplicate Account Detection


Problem Statement

Create a function to detect duplicate usernames in a system.


```python
# Function to detect duplicate usernames
def detect_duplicates(usernames):
    
    # Convert list to set
    unique_usernames = set(usernames)
    
    # If length reduces → duplicates exist
    if len(unique_usernames) < len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Example username list
accounts = ["user1", "user2", "user3", "user1"]

# Call function
detect_duplicates(accounts)
```

# Problem 6: Agriculture – Premium Crop Price Filter


Problem Statement

Create a function that filters crop prices above a premium threshold.


```python
# Function to filter premium crop prices
def filter_premium_crops(prices):
    
    premium_crops = []
    
    # Loop through price list
    for price in prices:
        
        # Check if price is above 2000
        if price > 2000:
            premium_crops.append(price)
    
    print("Premium Crops:", premium_crops)


# Example crop prices
crop_prices = [1500, 2500, 1800, 3200, 1900]

# Call function
filter_premium_crops(crop_prices)
```

# Problem 9: E-Learning – Student Attendance Eligibility System


Problem Statement

Create a function to calculate attendance percentage and check exam eligibility.


```python
# Function to calculate attendance and check exam eligibility
def check_exam_eligibility(attendance_list):
    
    present_count = 0
    
    # Step 1: Count present days using loop
    for status in attendance_list:
        if status == "P":
            present_count += 1
    
    # Step 2: Calculate attendance percentage
    total_days = len(attendance_list)
    percentage = (present_count / total_days) * 100
    
    print("Attendance Percentage:", percentage)
    
    # Step 3: Check eligibility
    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")


# Example attendance data
attendance = ["P", "P", "A", "P", "P"]

# Call function
check_exam_eligibility(attendance)
```

# Problem 10: Smart Electricity Bill Analyzer


Problem Statement

You are building a smart electricity billing system for residential users.
Create a function that calculates the electricity bill and classifies usage level based on units consumed.

```python
# Function to calculate electricity bill and usage classification
def calculate_electricity_bill(units):

    bill = 0

    # Step 1: Calculate bill based on slab rates
    if units <= 100:
        bill = units * 3

    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5

    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7

    # Step 2: Classify usage level
    if bill < 500:
        status = "Low Usage"

    elif bill <= 1500:
        status = "Moderate Usage"

    else:
        status = "High Usage"

    return bill, status


# Example usage
units_consumed = 250

total_bill, usage_status = calculate_electricity_bill(units_consumed)

print("Units Consumed:", units_consumed)
print("Total Bill: ₹", total_bill)
print("Usage Classification:", usage_status)
```
