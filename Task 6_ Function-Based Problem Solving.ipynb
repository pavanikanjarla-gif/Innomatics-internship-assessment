# Problem Statement 1:Smart Parking Lot Management System


Design a function to manage a smart parking lot.

The system should:
Accept vehicle entry and exit logs


Calculate total parked vehicles


Identify peak parking usage


Alert if parking exceeds capacity


```python
# Function to manage smart parking lot
def manage_parking(capacity, vehicle_logs):
    
    current_vehicles = 0
    peak_usage = 0
    
    # Step 1: Process vehicle logs
    for log in vehicle_logs:
        
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            current_vehicles -= 1
        
        # Track peak usage
        if current_vehicles > peak_usage:
            peak_usage = current_vehicles
    
    print("Currently Parked Vehicles:", current_vehicles)
    
    # Step 2: Check parking status
    if current_vehicles > capacity:
        print("Parking Status: Exceeded Capacity! Alert!")
    else:
        print("Parking Status: Available")
    
    # Optional: Show peak usage
    print("Peak Parking Usage:", peak_usage)


# Sample Input
parking_capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

# Call function
manage_parking(parking_capacity, vehicle_logs)
```

# Problem Statement 2:Online Food Delivery Time Estimator


Create a function that estimates delivery time based on:

Distance (km)


Weather condition


Traffic level


Apply delays dynamically and display final ETA.


```python
# Function to estimate delivery time
def estimate_delivery_time(distance, traffic, weather):

    total_time = 5   # base preparation time

    # Travel time
    total_time += distance * 5

    # Traffic delay
    if traffic.lower() == "high":
        total_time += 5
    elif traffic.lower() == "medium":
        total_time += 3
    elif traffic.lower() == "low":
        total_time += 1

    # Weather delay
    if weather.lower() == "rainy":
        total_time += 5
    elif weather.lower() == "stormy":
        total_time += 10

    print("Estimated Delivery Time:", total_time, "minutes")


# Sample Input
distance_km = 8
traffic_level = "High"
weather_condition = "Rainy"

estimate_delivery_time(distance_km, traffic_level, weather_condition)
```

# Problem Statement 3:Movie Theatre Seat Occupancy Analyzer

Build a function that analyzes seat booking data and:

Calculates occupancy percentage


Determines if show is Housefull


Suggests opening additional shows


```python
# Function to analyze seat occupancy
def analyze_seat_occupancy(total_seats, booked_seats):
    
    # Step 1: Count booked seats using loop
    booked_count = 0
    
    for seat in booked_seats:
        if seat == 1:
            booked_count += 1
    
    # Step 2: Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100
    
    print("Occupancy:", int(occupancy), "%")
    
    # Step 3: Determine show status
    if occupancy == 100:
        print("Show Status: Housefull")
    
    elif occupancy >= 75:
        print("Show Status: Almost Full")
        print("Suggestion: Consider opening additional shows")
    
    elif occupancy >= 50:
        print("Show Status: Moderate Occupancy")
    
    else:
        print("Show Status: Low Occupancy")


# Sample Input
total_seats = 200
booked_seats = [1] * 150   # 150 seats booked

# Call function
analyze_seat_occupancy(total_seats, booked_seats)
```

# Problem Statement 4:Cloud Server Load Classification System


Create a function to classify server load based on CPU usage readings.
    
Rules:
Average CPU < 50% → Normal


50%–80% → Warning


80% → Critical



```python
# Function to classify server load
def classify_server_load(cpu_readings):
    
    total = 0
    
    # Step 1: Calculate total using loop
    for value in cpu_readings:
        total += value
    
    # Step 2: Calculate average
    average = total / len(cpu_readings)
    
    # Print average exactly as required
    print("Average CPU Load:", int(average), "%")
    
    # Step 3: Classify server status
    if average < 50:
        print("Server Status: Normal")
    
    elif average <= 80:
        print("Server Status: Warning")
    
    else:
        print("Server Status: Critical")


# Sample Input
cpu_data = [45, 60, 70, 85, 90]

# Call function
classify_server_load(cpu_data)
```

# Problem Statement 5:Smart Classroom Resource Usage Monitor


Design a function that tracks usage of classroom resources (projector, AC, lights) and identifies overuse patterns.


```python
# Function to monitor classroom resource usage
def monitor_resources(resource_usage):
    
    overused = []
    
    # Step 1: Check each resource using loop
    for resource, hours in resource_usage.items():
        
        # Assume usage above 8 hours is overuse
        if hours > 8:
            overused.append(resource)
    
    # Step 2: Display results
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
usage_data = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

# Call function
monitor_resources(usage_data)
```

#  Problem Statement 6: Online Event Registration Capacity Controller


Create a function that manages event registrations by:

Tracking registrations


Preventing overbooking


Triggering waitlist mode


```python
# Function to manage event registrations
def manage_registrations(capacity, total_registrations):
    
    confirmed = 0
    waitlisted = 0
    
    # Step 1: Process registrations using loop
    for i in range(total_registrations):
        
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1
    
    # Step 2: Print results
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    
    if waitlisted > 0:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")


# Sample Input
event_capacity = 100
registrations = 105

# Call function
manage_registrations(event_capacity, registrations)
```
