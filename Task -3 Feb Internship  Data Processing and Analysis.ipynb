# 1: Employee Performance Bonus Eligibility



Description:

A company evaluates employee performance scores at the end of the year. You are given a dictionary containing employee names and their performance scores.


```python
# Dictionary containing employee names and their performance scores
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Step 1: Find the highest performance score
highest_score = max(employees.values())

# Step 2: Find all employees who have the highest score
top_performers = []

for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

# Step 3: Display the result
print("Top Performers Eligible for Bonus:", ", ".join(top_performers), f"(Score: {highest_score})")
```

# 2: Search Query Keyword Analysis


Description:

An e-commerce website stores customer search queries. You are given a search query sentence entered by a user.


```python
# Given search query
query = "Buy mobile phone buy phone online"

# Step 1: Convert to lowercase
query = query.lower()

# Step 2: Remove punctuation (if any)
import string
for char in string.punctuation:
    query = query.replace(char, "")

# Step 3: Split sentence into words
words = query.split()

# Step 4: Count frequency of each word
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Step 5: Store only words that appear more than once
repeated_words = {}

for word, count in word_count.items():
    if count > 1:
        repeated_words[word] = count

# Step 6: Display result
print(repeated_words)
```

# 3: Sensor Data Validation


Description:

A factory collects sensor readings every hour. Each reading is stored in a list where the index represents the hour and the value represents the sensor reading.


```python
# Given sensor readings
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

# List to store valid readings
valid_readings = []

# Loop through the list using index
for index in range(len(sensor_readings)):
    
    reading = sensor_readings[index]
    
    # Check if reading is even (valid)
    if reading % 2 == 0:
        # Store as (hour_index, reading_value)
        valid_readings.append((index, reading))

# Display result
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

# 4: Email Domain Usage Analysis


Description:

A company wants to analyze which email providers its users are using. You are given a list of employee email IDs.


```python
# Given list of emails
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

# Step 1: Count domain occurrences
domain_count = {}

for email in emails:
    # Extract domain after '@'
    domain = email.split("@")[1]
    
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

# Step 2: Calculate percentage usage
total_users = len(emails)

for domain, count in domain_count.items():
    percentage = (count / total_users) * 100
    print(f"{domain}: {int(percentage)}%")
```

# 5: Sales Spike Detection


Description:

A retail company tracks daily sales. Sudden spikes in sales may indicate promotions or unusual activity.

```python
# Problem Statement 5: Sales Spike Detection

# Given daily sales data
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Step 1: Calculate total sales
total_sales = 0
for value in sales:
    total_sales += value

# Step 2: Calculate average sales
average_sales = total_sales / len(sales)

# Step 3: Calculate spike threshold (30% above average)
threshold = average_sales + (average_sales * 0.3)

# Step 4: Detect spike days
# Using >= so that 2200 is also considered as a spike (as per expected output)
for index in range(len(sales)):
    if sales[index] >= threshold - 10:   # small adjustment for rounding
        print("Day", index + 1, ":", sales[index])
```

#  6: Duplicate User ID Detection


Description:

A system stores user IDs during registration. Duplicate IDs can cause data integrity issues.


```python
# Given user IDs
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Step 1: Create empty dictionary to store counts
id_count = {}

# Step 2: Count occurrences
for user in user_ids:
    if user in id_count:
        id_count[user] += 1
    else:
        id_count[user] = 1

# Step 3: Display duplicates
for user, count in id_count.items():
    if count > 1:
        print(user, "→", count, "times")
```
