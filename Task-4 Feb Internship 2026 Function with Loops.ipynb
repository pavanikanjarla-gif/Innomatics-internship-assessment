# Problem 1: Mobile Recharge Validation System

Problem Statement:
Create a function to validate a mobile recharge request.

Rules:
Recharge amount must be ≥ ₹50

Valid plans: 199, 299, 399, 599

User should be allowed to retry if invalid


```python
# Function to validate recharge amount
def validate_recharge():
    
    valid_plans = [199, 299, 399, 599]
    
    while True:
        try:
            amount = int(input("Enter recharge amount: ₹"))
            
            # Rule 1: Minimum recharge amount
            if amount < 50:
                print("Recharge amount must be at least ₹50.")
                print("Please try again.\n")
                continue
            
            # Rule 2: Check if plan is valid
            if amount in valid_plans:
                print("Recharge successful! Plan activated.")
                break
            else:
                print("Invalid plan selected.")
                print("Available plans are:", valid_plans)
                print("Please try again.\n")
        
        except ValueError:
            print("Invalid input! Please enter a numeric value.\n")


# Call the function
validate_recharge()
```

# Problem 2: Inventory Reorder Alert System


Problem Statement:

Given a dictionary of products and stock quantities:

Print Reorder Alert if stock < 15


Print Stock OK otherwise




```python
# Function to check stock levels
def check_inventory(products):
    
    for item, quantity in products.items():
        
        if quantity < 15:
            print(item, "→ Reorder Alert (Stock:", quantity, ")")
        else:
            print(item, "→ Stock OK (Stock:", quantity, ")")


# Given product stock dictionary
inventory = {
    "Rice": 20,
    "Sugar": 10,
    "Oil": 5,
    "Wheat": 18
}

# Call the function
check_inventory(inventory)
```

#  Problem 3: Student Result Processing System


Problem Statement

Create a function that:

Accepts a list of student marks


Calculates average marks


Prints Pass if average ≥ 50 else Fail


```python
# Function to process student result
def process_result(marks):
    
    total = 0
    
    # Step 1: Calculate total marks using loop
    for mark in marks:
        total += mark
    
    # Step 2: Calculate average
    average = total / len(marks)
    
    # Step 3: Check pass or fail
    if average >= 50:
        return "Pass"
    else:
        return "Fail"


# Given list of student marks
student_marks = [60, 45, 70, 55]

# Call function and print result
result = process_result(student_marks)

print("Average Marks:", sum(student_marks) / len(student_marks))
print("Result:", result)
```

# Problem 4: Cab Fare Estimator with Retry Option

Problem Statement

Create a cab fare calculator:

Base fare = ₹50


₹12 per km


Peak hour → 25% extra


Allow retry using while loop


# Function to calculate fare
def calculate_fare(distance, peak):
    
    base_fare = 50
    per_km_rate = 12
    
    # Step 1: Calculate normal fare
    fare = base_fare + (distance * per_km_rate)
    
    # Step 2: Add peak hour charge if applicable
    if peak.lower() == "yes":
        fare = fare + (fare * 0.25)   # 25% extra
    
    return fare


# While loop for retry option
while True:
    
    try:
        distance = float(input("Enter distance travelled (in km): "))
        peak = input("Is it peak hour? (yes/no): ")
        
        total_fare = calculate_fare(distance, peak)
        
        print("Total Fare: ₹", total_fare)
        
        retry = input("Do you want to calculate again? (yes/no): ")
        
        if retry.lower() != "yes":
            print("Thank you for using Cab Fare Estimator.")
            break
    
    except ValueError:
        print("Invalid input! Please enter numeric value for distance.\n")


# Problem 5: Employee Attendance Eligibility Checker


Problem Statement
Create a function that:

Accepts attendance list ("P", "A")


Calculates attendance percentage


Returns Eligible if ≥ 75% else Not Eligible


```python
# Function to check attendance eligibility
def check_attendance(attendance_list):
    
    present_days = 0
    
    # Step 1: Count present days using loop
    for status in attendance_list:
        if status == "P":
            present_days += 1
    
    # Step 2: Calculate attendance percentage
    total_days = len(attendance_list)
    percentage = (present_days / total_days) * 100
    
    # Step 3: Check eligibility
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# Example attendance list
attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]

# Call function
result = check_attendance(attendance)

print("Attendance Percentage:", (attendance.count("P") / len(attendance)) * 100)
print("Eligibility Status:", result)
```

# Problem 6: Password Strength Checker


Problem Statement
Create a function to check password strength.

Rules:
Minimum length = 8


Must contain at least one digit


Must contain at least one special character (@#$)


```python
# Function to check password strength
def check_password(password):
    
    has_digit = False
    has_special = False
    
    special_characters = "@#$"
    
    # Rule 1: Minimum length
    if len(password) < 8:
        return "Weak Password (Minimum length should be 8)"
    
    # Rule 2 & 3: Check digit and special character
    for char in password:
        
        if char.isdigit():
            has_digit = True
        
        if char in special_characters:
            has_special = True
    
    # Final strength check
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password (Must contain at least one digit and one special character)"
    

# Example usage
user_password = input("Enter your password: ")

result = check_password(user_password)

print(result)
```
